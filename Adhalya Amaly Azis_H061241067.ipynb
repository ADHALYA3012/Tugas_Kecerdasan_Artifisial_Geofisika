## **PENERAPAN PROMPT SEBAGAI WORK SPECIFICATION PADA KASUS GEOFISIKA**

![Screenshot 2026-09-28 204358.png](<attachment:Screenshot 2026-09-28 204358.png>)

Dalam kegiatan geofisika, kecerdasan artifisial dapat digunakan sebagai alat bantu untuk membaca data, melakukan analisis, membuat visualisasi, dan membantu interpretasi. Namun, agar hasil yang diberikan AI sesuai dengan kebutuhan ilmiah, instruksi yang diberikan perlu disusun secara jelas. Dalam Chapter 2 *AI Cookbook untuk Earth Sciences*, prompt dipandang sebagai work specification, yaitu spesifikasi pekerjaan yang menjelaskan tujuan, konteks, input, batasan, output, kriteria kualitas, dan pemeriksaan. 

Struktur prompt dapat disusun melalui komponen Role, Task, Context, Input, Constraints, Output, Quality Criteria, dan Verification (R–T–C–I–K–O–Q–V) yang mementingkan kejelasan, relevansi, konsistensi, dan sejauh mana instruksi tersebut mengurangi keputusan penting yang harus ditebak oleh AI. Digunakan juga command verb yang sesuai dengan tujuan, misalnya analisis untuk critical thinking, hitung untuk problem solving, dan rancang untuk pembuatan artifact.

### **KASUS 1 — INTERPRETASI PENAMPANG SEISMIK 2D**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang belajar melakukan interpretasi dasar data seismik refleksi.

#### **Task**

Analisis penampang seismik 2D untuk mengidentifikasi reflektor utama, memperkirakan batas antar-lapisan berdasarkan pola reflektor, serta mengidentifikasi kemungkinan struktur geologi yang terlihat pada penampang.

#### **Context**

Data seismik refleksi digunakan untuk memperoleh gambaran kondisi bawah permukaan berdasarkan pantulan gelombang seismik. Pantulan dapat muncul ketika gelombang melewati batas material yang memiliki perbedaan sifat fisik.

Pada penampang seismik, interpretasi dilakukan dengan memperhatikan pola reflektor, antara lain kontinuitas, bentuk, kemiringan, perubahan amplitudo, dan posisi terhadap waktu tempuh dua arah (*Two-Way Travel Time* atau TWT).

Interpretasi dalam kasus ini merupakan interpretasi awal berdasarkan karakter yang terlihat pada penampang, sehingga tidak semua reflektor dapat langsung dikaitkan dengan jenis batuan atau struktur tertentu.

#### **Input**

Data yang digunakan berupa:

* Penampang seismik 2D dalam bentuk gambar;
* Sumbu horizontal berupa posisi/jarak sepanjang lintasan;
* Sumbu vertikal berupa TWT;
* Informasi tambahan mengenai lokasi atau lintasan jika tersedia.

#### **Constraints**

1. Jangan menentukan jenis batuan hanya berdasarkan bentuk atau amplitudo reflektor.
2. Gunakan karakter yang benar-benar terlihat pada penampang sebagai dasar interpretasi.
3. Bedakan antara pengamatan langsung dan interpretasi.
4. Jangan menyatakan suatu struktur sebagai pasti jika bukti pada penampang tidak cukup.
5. Jangan mengubah atau menghilangkan bagian data.
6. Jika reflektor tidak jelas, nyatakan sebagai bagian yang memiliki ketidakpastian.
7. Jangan mengubah TWT menjadi kedalaman apabila kecepatan gelombang atau model kecepatan tidak tersedia.

#### **Output**

Hasil yang direncanakan berupa:

1. identifikasi reflektor utama
2. deskripsi kontinuitas dan bentuk reflektor
3. perkiraan batas lapisan berdasarkan perubahan pola reflektor
4. identifikasi kemungkinan struktur geologi, jika terdapat indikasi
5. alasan yang mendukung setiap interpretasi
6. tabel yang memisahkan pengamatan dan interpretasi
7. kesimpulan singkat mengenai pola bawah permukaan.

#### **Quality Criteria**

Hasil harus:

* Berdasarkan pola yang benar-benar terlihat pada penampang
* Mempertimbangkan kontinuitas reflektor
* Mempertimbangkan bentuk dan kemiringan reflektor
* Mempertimbangkan perubahan amplitudo jika terlihat
* Mencantumkan posisi TWT yang relevan
* Membedakan fakta pengamatan dari dugaan interpretasi
* Menyatakan ketidakpastian apabila bukti tidak cukup

#### **Verification**

Periksa kembali seluruh penampang setelah interpretasi dilakukan. Pastikan setiap reflektor dan struktur yang disebutkan memiliki dasar visual pada data. Periksa pula apakah interpretasi konsisten dengan pola reflektor di bagian penampang yang berdekatan. Jika suatu interpretasi tidak dapat diverifikasi dari penampang, tandai sebagai tidak pasti dan jangan menyajikannya sebagai kesimpulan pasti.

#### **Rencana hasil**

Output yang diharapkan bukan berupa penentuan litologi secara pasti, melainkan peta atau deskripsi reflektor utama, kemungkinan batas lapisan, dan kemungkinan struktur berdasarkan bukti yang terlihat pada penampang.

### **KASUS 2 — ANALISIS DATA GEOLISTRIK RESISTIVITAS**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang belajar mengolah dan menginterpretasikan data geolistrik resistivitas 2D.

#### **Task**

Analisis data resistivitas 2D untuk mengidentifikasi zona dengan nilai resistivitas relatif rendah dan tinggi serta melihat kemungkinan batas atau perubahan kondisi bawah permukaan berdasarkan distribusi resistivitas.

#### **Context**

Metode geolistrik resistivitas digunakan untuk memperoleh informasi mengenai variasi sifat listrik bawah permukaan. Hasil pengolahan dapat ditampilkan sebagai penampang resistivitas 2D. Perbedaan resistivitas dapat menunjukkan adanya perubahan material atau kondisi bawah permukaan, tetapi nilai resistivitas tidak selalu memiliki hubungan satu-ke-satu dengan jenis batuan tertentu. Interpretasi harus mempertimbangkan konteks geologi dan keterbatasan data.

#### **Input**

Input terdiri atas:

* Data resistivitas hasil pengukuran atau hasil inversi
* Posisi elektroda atau jarak sepanjang lintasan
* Kedalaman atau pseudo-depth sesuai format data
* Nilai resistivitas
* Jika tersedia, informasi topografi dan kondisi geologi daerah penelitian.

#### **Constraints**

1. Jangan langsung menyatakan suatu nilai resistivitas sebagai jenis batuan tertentu.
2. Jangan mengubah data resistivitas asli.
3. Bedakan resistivitas hasil pengukuran dari hasil interpretasi.
4. Jangan menyatakan keberadaan objek geologi tertentu jika tidak terdapat pola resistivitas yang mendukung.
5. Jika data berupa hasil inversi, jangan menganggap model inversi sebagai representasi pasti kondisi bawah permukaan.
6. Perhatikan satuan resistivitas.
7. Jika informasi kedalaman sebenarnya tidak tersedia, jangan mengubah pseudo-depth menjadi kedalaman sebenarnya.

#### **Output**

Hasil yang direncanakan berupa:

1. identifikasi zona resistivitas rendah
2. identifikasi zona resistivitas tinggi
3. deskripsi perubahan lateral dan vertikal resistivitas
4. kemungkinan batas antar-zona
5. kemungkinan interpretasi geofisika berdasarkan pola resistivitas
6. tabel zona, posisi, nilai resistivitas, pengamatan, dan interpretasi
7. kesimpulan mengenai pola resistivitas bawah permukaan.

#### **Quality Criteria**

Analisis harus:

* Menggunakan nilai resistivitas dari data
* Mempertahankan satuan
* Membedakan observasi dan interpretasi
* Memperhatikan posisi spasial
* Tidak menghubungkan nilai resistivitas dengan litologi secara berlebihan
* Menjelaskan dasar dari setiap interpretasi.

#### **Verification**

Periksa kembali penampang resistivitas secara keseluruhan. Pastikan zona yang disebut sebagai resistivitas tinggi atau rendah memang sesuai dengan rentang nilai pada data. Periksa pula apakah interpretasi suatu zona konsisten dengan pola di sekitarnya dan apakah terdapat bagian yang memiliki resolusi atau kualitas data yang rendah.

#### **Rencana hasil**

Output yang diharapkan berupa penampang yang telah diidentifikasi zona resistivitasnya, tabel karakteristik setiap zona, serta interpretasi awal mengenai kemungkinan perubahan kondisi bawah permukaan.

### **KASUS 3 — ANALISIS HUBUNGAN CURAH HUJAN DAN NDVI**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari hubungan antara variabel meteorologi dan kondisi vegetasi menggunakan analisis data.

#### **Task**

Analisis hubungan temporal antara curah hujan dan NDVI untuk mengetahui pola perubahan vegetasi terhadap variasi curah hujan serta mengidentifikasi apakah hubungan tersebut menunjukkan pola linear atau non-linear.

#### **Context**

Curah hujan merupakan salah satu faktor yang dapat memengaruhi ketersediaan air bagi vegetasi. NDVI dapat digunakan sebagai indikator kondisi atau tingkat kehijauan vegetasi. Hubungan antara curah hujan dan NDVI tidak harus selalu linear. Respons vegetasi dapat berbeda pada kondisi curah hujan rendah, sedang, dan tinggi. Oleh karena itu, analisis perlu melihat pola data terlebih dahulu sebelum menentukan model yang sesuai.

#### **Input**

Input berupa:

* Data curah hujan bulanan
* Data NDVI bulanan
* Periode pengamatan
* Lokasi atau wilayah pengamatan
* jika tersedia, informasi musim atau variabel meteorologi tambahan.

#### **Constraints**

1. Jangan menyatakan bahwa curah hujan adalah satu-satunya faktor yang menentukan NDVI.
2. Jangan menganggap hubungan selalu linear sebelum melihat pola data.
3. Pertahankan satuan curah hujan dan periode waktu.
4. Jangan mengisi data yang hilang secara otomatis tanpa menjelaskan metode.
5. Bedakan korelasi dari hubungan sebab-akibat.
6. Jika terdapat keterlambatan respons vegetasi, jangan mengabaikannya.
7. Jangan menyatakan kekeringan hanya berdasarkan NDVI atau curah hujan tanpa menjelaskan indikator yang digunakan.

#### **Output**

Hasil yang direncanakan berupa:

1. Grafik time series curah hujan dan NDVI
2. Scatter plot curah hujan terhadap NDVI
3. Identifikasi pola hubungan
4. Analisis apakah hubungan cenderung linear atau non-linear
5. Jika sesuai, perbandingan model linear dan model non-linear
6. Nilai statistik atau ukuran kecocokan model yang digunakan
7. Interpretasi hubungan meteorologi–vegetasi
8. Penjelasan mengenai kemungkinan lag antara curah hujan dan respons NDVI.

#### **Quality Criteria**

Analisis harus:

* Menggunakan periode yang sama untuk kedua variabel
* Mencantumkan satuan
* Membedakan korelasi dan kausalitas
* Mempertimbangkan pola musiman
* Tidak memilih model hanya berdasarkan asumsi
* Menjelaskan alasan pemilihan model
* Menyatakan keterbatasan analisis.

#### **Verification**

Periksa kembali kesesuaian tanggal atau periode antara curah hujan dan NDVI. Periksa missing value, data ganda, satuan, serta rentang nilai.

Jika model digunakan, bandingkan hasil model dengan pola data aktual dan periksa apakah kesimpulan yang dibuat benar-benar didukung oleh data.

#### **Rencana hasil**

Output yang diharapkan adalah gambaran hubungan temporal antara curah hujan dan NDVI, termasuk kemungkinan pola non-linear atau lag, bukan kesimpulan bahwa perubahan NDVI hanya disebabkan oleh curah hujan.

### **KASUS 4 — ANALISIS POLA ARUS LAUT**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari analisis data oseanografi, khususnya pola arus laut.

#### **Task**

Analisis data komponen arus laut untuk menggambarkan arah, kecepatan, dan pola spasial arus serta mengidentifikasi kemungkinan pola konvergensi, divergensi, atau sirkulasi apabila terdapat bukti pada data.

#### **Context**

Arus laut dapat direpresentasikan menggunakan komponen kecepatan, misalnya komponen zonal dan meridional. Kedua komponen tersebut dapat digunakan untuk menentukan arah dan besar kecepatan arus. Visualisasi seperti *quiver* atau *streamline* dapat digunakan untuk membantu melihat pola spasial medan arus. Namun, pola visualisasi harus tetap dibedakan dari interpretasi dinamika laut yang lebih spesifik.

#### **Input**

Input berupa:

* Komponen arus zonal
* Komponen arus meridional
* Koordinat longitude dan latitude
* Waktu pengamatan
* Jika tersedia, informasi kedalaman
* Konfigurasi atau periode data yang digunakan.

#### **Constraints**

1. Jangan menentukan penyebab dinamika arus hanya berdasarkan arah panah.
2. Jangan menyatakan adanya pusaran (*eddy*) jika pola sirkulasinya belum jelas.
3. Jangan mengubah koordinat atau sistem satuan.
4. Bedakan arah arus dari interpretasi dinamika laut.
5. Jika data hanya menunjukkan kondisi pada satu waktu, jangan menyimpulkan pola temporal.
6. Jangan menginterpretasikan pola sebagai arus permukaan jika kedalaman data tidak diketahui.
7. Jangan menghilangkan nilai kosong tanpa penjelasan.

#### **Output**

Hasil yang direncanakan berupa:

1. Peta vektor arus
2. Visualisasi *streamline* jika sesuai
3. Informasi arah dominan arus
4. Distribusi relatif kecepatan
5. Identifikasi kemungkinan pola sirkulasi
6. Lokasi yang menunjukkan perubahan arah atau kecepatan yang menonjol
7. Interpretasi oseanografis berdasarkan pola data
8. Kesimpulan singkat mengenai pola arus.

#### **Quality Criteria**

Hasil harus:

* Mempertahankan koordinat geografis
* Menggunakan satuan kecepatan yang konsisten
* Membedakan arah dan besar kecepatan
* Menjelaskan pola berdasarkan data
* Tidak memberikan interpretasi dinamika yang tidak didukung data
* Mencantumkan periode dan kedalaman data jika tersedia.

#### **Verification**

Periksa apakah arah vektor sesuai dengan komponen zonal dan meridional. Periksa pula apakah skala visualisasi tidak membuat arus terlihat lebih kuat atau lebih lemah secara menyesatkan. Jika terdapat pola sirkulasi, periksa kembali pola tersebut pada keseluruhan domain sebelum menyebutkannya sebagai interpretasi.

#### **Rencana hasil**

Output yang diharapkan berupa peta pola arus yang menunjukkan arah dan kecepatan, disertai interpretasi mengenai pola spasial yang terlihat. Interpretasi mengenai mekanisme pembentukan arus hanya diberikan jika terdapat data pendukung yang cukup.


### **KASUS 5 — ANALISIS PERUBAHAN TUTUPAN LAHAN MENGGUNAKAN CITRA SATELIT**

#### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari analisis geospasial dan penginderaan jauh menggunakan data citra satelit.

#### **Task**

Analisis citra satelit multitemporal untuk mengidentifikasi perubahan tutupan lahan pada wilayah studi dan menghasilkan informasi spasial mengenai lokasi serta jenis perubahan yang terdeteksi.

#### **Context**

Citra satelit menyediakan informasi permukaan bumi dalam bentuk data spasial dan spektral. Citra dari waktu yang berbeda dapat dibandingkan untuk melihat perubahan kondisi permukaan. Analisis perubahan tutupan lahan perlu memperhatikan kesesuaian waktu akuisisi, sistem koordinat, resolusi spasial, kualitas citra, dan metode klasifikasi atau deteksi perubahan yang digunakan.

#### **Input**

Input berupa:

* Citra satelit periode pertama
* Citra satelit periode kedua
* Batas wilayah studi
* Metadata citra
* Informasi tanggal akuisisi
* Sistem koordinat
* Jika tersedia, data referensi atau *ground truth*.

#### **Constraints**

1. Jangan membandingkan citra tanpa memeriksa kesesuaian sistem koordinat.
2. Jangan menganggap perbedaan nilai piksel sebagai perubahan tutupan lahan sebelum memeriksa faktor lain.
3. Pertahankan informasi tanggal akuisisi.
4. Perhatikan perbedaan kondisi atmosfer atau kualitas citra.
5. Jangan menentukan kelas tutupan lahan yang tidak dapat dibedakan dari data.
6. Jangan mengubah data asli.
7. Jika tidak tersedia data referensi, nyatakan keterbatasan validasi klasifikasi.
8. Bedakan perubahan yang terdeteksi dari penyebab perubahan.

#### **Output**

Hasil yang direncanakan berupa:

1. Peta tutupan lahan periode pertama
2. Peta tutupan lahan periode kedua
3. Peta perubahan
4. Luas atau proporsi setiap kelas jika perhitungan memungkinkan
5. Tabel perubahan antar-kelas
6. Lokasi perubahan yang paling menonjol
7. Statistik akurasi jika tersedia data validasi
8. Interpretasi spasial terhadap pola perubahan
9. Kesimpulan mengenai perubahan tutupan lahan.

#### **Quality Criteria**

Hasil harus:

* Menggunakan sistem koordinat yang jelas
* Mencantumkan tanggal citra
* Mempertimbangkan resolusi spasial
* Menjelaskan metode klasifikasi atau deteksi perubahan
* Membedakan perubahan nyata dari kemungkinan artefak data
* Tidak memberikan klaim akurasi tanpa data validasi
* Mempertahankan informasi provenance citra.

### **Verification**

Periksa kembali apakah kedua citra memiliki sistem koordinat dan cakupan wilayah yang sesuai. Periksa hasil klasifikasi atau deteksi perubahan terhadap citra asli. Jika data validasi tersedia, bandingkan hasil klasifikasi dengan data referensi. Jika tidak tersedia, nyatakan bahwa hasil belum memperoleh validasi independen.

### **Rencana hasil**

Output yang diharapkan berupa dua peta kondisi tutupan lahan dan satu peta perubahan, dilengkapi tabel statistik serta interpretasi spasial mengenai lokasi perubahan. Hasil tidak secara otomatis digunakan untuk menentukan penyebab perubahan tanpa data tambahan.
